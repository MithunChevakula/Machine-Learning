```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA

X=np.array([[-1,1],[-2,-1],[-3,-2],[1,1],[2,1],[3,2]])

pca=PCA(n_components=2)
pca.fit(X)

print(pca.explained_variance_ratio_)
print(pca.singular_values_)
```
